```python
import random
import time
import matplotlib.pyplot as plt
```

```python
# All lists
arr_100_sorted = [i for i in range(100)]
arr_200_sorted = [i for i in range(200)]
arr_300_sorted = [i for i in range(300)]
arr_400_sorted = [i for i in range(400)]
arr_500_sorted = [i for i in range(500)]
arr_1000_sorted = [i for i in range(1000)]

arr_100_reverse_sorted = [i for i in range(100, 0, -1)]
arr_200_reverse_sorted = [i for i in range(200, 0, -1)]
arr_300_reverse_sorted = [i for i in range(300, 0, -1)]
arr_400_reverse_sorted = [i for i in range(400, 0, -1)]
arr_500_reverse_sorted = [i for i in range(500, 0, -1)]
arr_1000_reverse_sorted = [i for i in range(1000, 0, -1)]

arr_100_random = [random.randint(0, 100) for i in range(100)]
arr_200_random = [random.randint(0, 200) for i in range(200)]
arr_300_random = [random.randint(0, 300) for i in range(300)]
arr_400_random = [random.randint(0, 400) for i in range(400)]
arr_500_random = [random.randint(0, 500) for i in range(500)]
arr_1000_random = [random.randint(0, 1000) for i in range(1000)]
```

```python
#bubble sort
def bubble_sort(arr):
  arr_len = len(arr)
  for i in range(0, arr_len-1):
    for j in range(0, arr_len-1-i):
      if arr[j+1] < arr[j]:
        arr[j], arr[j+1] = arr[j+1], arr[j]
  return arr


```

```python
#Selection Sort

def selection_sort(arr):

  for i in range(len(arr) - 1):
    min_index = i
    for j in range(i+1, len(arr)):
      if arr[j] < arr[min_index]:
        min_index = j
    arr[i], arr[min_index] = arr[min_index], arr[i]
  return arr



```

```python
#Insertion Sort

def insertion_sort(arr):
  for i in range(1, len(arr)):
    key = arr[i]
    j = i-1
    while j >= 0 and key < arr[j]:
      arr[j+1] = arr[j]
      j -= 1
      arr[j+1] = key

  return(arr)

```

Bubble sort with each array type

```python
#execution times bubble sort (sorted data)
sorted_arr_list = [arr_100_sorted.copy(),arr_200_sorted.copy(),arr_300_sorted.copy(), arr_400_sorted.copy(), arr_500_sorted.copy(),arr_1000_sorted.copy()]
execution_times_bubble_sorted = []

for i in range(6):
  start_time = time.perf_counter()
  bubble_sort(sorted_arr_list[i])
  end_time = time.perf_counter()
  execution_time = end_time - start_time
  execution_times_bubble_sorted.append(execution_time)
```
